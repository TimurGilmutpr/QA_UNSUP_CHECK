```python
import os
import pandas as pd
```

```python
data_lsit = [k for k in os.listdir() if k.endswith("xlsx")]
```

```python
dfs_all = []
for k in data_lsit:
    dfs_all.append(pd.read_excel(k))
df = pd.concat(dfs_all, ignore_index=True)
df.head()
```
