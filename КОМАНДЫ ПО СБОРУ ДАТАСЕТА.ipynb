```python
sudo apt update
sudo apt install poppler-utils   # для pdftotext
sudo apt install catdoc          # для .doc (Word 97-2003)
```

```python
# Создадим пустой файл для датасета
> dataset.txt

# Извлечь текст из всех PDF
for f in *.pdf; do
    echo "=== $f ===" >> dataset.txt
    pdftotext "$f" - >> dataset.txt
    echo -e "\n\n" >> dataset.txt
done

# Извлечь текст из всех .doc
for f in *.doc; do
    echo "=== $f ===" >> dataset.txt
    catdoc "$f" >> dataset.txt
    echo -e "\n\n" >> dataset.txt
done

# Если есть .ipynb — можно пропустить (или извлечь через jupyter nbconvert)
# Но у тебя только Untitled.ipynb (скорее всего пустой или черновик)
```
